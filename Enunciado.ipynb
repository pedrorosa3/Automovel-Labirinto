## Introdução

Imagine um automóvel que deseja sair de um labirinto discreto. O automóvel (agente) ocupa uma célula e é direcional podendo estar orientado para Norte (N), Sul (S), Este (E) ou Oeste (O). Com uma única ação o agente pode mudar de orientação ou avançar com uma velocidade ajustável. 

O objetivo do automóvel é, partindo parado (com velocidade 0) de uma posição marcada e com uma determinada orientação, encontrar um plano para estacionar no quadrado de saída utilizando o menor número de ações possiveis. Estacionar o automóvel significa chegar à posição objetivo com velocidade zero.

## O labirinto

Considere que cada célula do mundo é dada por um par de coordenadas $(l,c)$ e que cada célula à esquerda de topo é a célula $(0,0)$ e que a célula do fundo à direita é $(M-1,N-1)$, com $M$ o número de linhas e $N$ o número de colunas do labirinto.

Na representação do labirinto considera-se que:
* o labirinto é cercado por paredes e pode conter obstáculos, cada um representado pelo símbolo "=";
* só há um carro, representado por "^", ">", "v" ou "<", consoante está virado para N, E, S ou O, respetivamente;
* só há um objetivo representado por "x";
* as casas livres são representadas por "."

Abaixo apresenta-se uma visualização de exemplo de um labirinto inicial de dimensão 7x7, onde o carro está na célula (5,1), e o objetivo na célula (1,1)
```python

= x . . . . =
= . . . = . =
= . . . = . =
= = = . = . =
= ^ . . . . =
= = = = = = =
```

## Objetivos

Formule este problema, como um problema de procura num grafo, de acordo com o Paradigma do Espaço de Estados, usando a implementação disponibilizada pelo módulo `searchPlus.py`. Deve minimizar a informação incluída no estado, formado apenas pelo que muda com as ações.

Assim, terá de completar a classe `Labirinto`:



````python
from searchPlus import *
import copy

line1 = "= = = = = = =\n"
line2 = "= x . . . . =\n"
line3 = "= . . . = . =\n"
line4 = "= . . . = . =\n"
line5 = "= = = . = . =\n"
line6 = "= ^ . . . . =\n"
line7 = "= = = = = = =\n"
grelha = line1 + line2 + line3 + line4 + line5 + line6 + line7

class Labirinto(Problem):

    def __init__(self, LabInicial=grelha, vmax=3):
        self.grelha=LabInicial;
        self.vmax=vmax;
        self.linhas=0;
        self.colunas=0;
        
        grelha=[];
        for i in self.grelha:
            if i=="\n":
                self.linhas=self.linhas+1
            if i!="\n" and i!=" ":
                grelha.append(i)
                self.colunas=self.colunas+1
        self.colunas=int(self.colunas/self.linhas)
                
        labirinto=[]
        
        for i in range(self.linhas):
            linha=grelha[i*self.colunas:(i+1)*self.colunas]
            labirinto.append(linha)
            
        for i in range(len(labirinto)):
            for j in range(len(labirinto[i])):
                if labirinto[i][j]=="x":
                    self.goal=[int(i),int(j)]
                if labirinto[i][j]=="^":
                    self.initial=[int(i),int(j),0,"N"]
                if labirinto[i][j]==">":
                    self.initial=[int(i),int(j),0,"E"]
                if labirinto[i][j]=="v":
                    self.initial=[int(i),int(j),0,"S"]
                if labirinto[i][j]=="<":
                    self.initial=[int(i),int(j),0,"O"]
            
        
        labirinto[self.initial[0]][self.initial[1]]="o" 
           
        self.labirinto=labirinto;
        
        

    def actions (self, state):
        
       
        listactions=[]
        
        if state[2]==0:
            listactions.append("E")
            listactions.append("D")




        
        
        if state[3]=="N":
            newvelo=state[2]+1
            if newvelo>self.vmax:
                newvelo=self.vmax
                
            flag=False
            for i in range(newvelo):
                if state[0]-(i+1)<0:
                    flag=True
                    break
                else:
                    celula=self.labirinto[state[0]-(i+1)][state[1]]
                    if celula == "=" :
                        flag=True
            if flag==False:
                listactions.append("A")


            
                    
        if state[3]=="E":
            
            newvelo=state[2]+1
            if newvelo>self.vmax:
                newvelo=self.vmax
                
            flag=False
            for i in range(newvelo):
                if state[1]+(i+1)>=self.colunas:
                    flag=True
                    break
                else:
                    celula=self.labirinto[state[0]][state[1]+(i+1)]
                    if celula == "=" :
                        flag=True
            if flag==False:
                listactions.append("A")
            
                

        if state[3]=="S":
            
            newvelo=state[2]+1
            if newvelo>self.vmax:
                newvelo=self.vmax
                
            flag=False
            for i in range(newvelo):
                if state[0]+(i+1)>=self.linhas:
                    flag=True
                    break
                else:
                    celula=self.labirinto[state[0]+(i+1)][state[1]]
                    if celula == "=" :
                        flag=True
            if flag==False:
                listactions.append("A")
                

        if state[3]=="O":
            
            newvelo=state[2]+1
            if newvelo>self.vmax:
                newvelo=self.vmax
                
            flag=False
            for i in range(newvelo):
                if state[1]-(i+1)<0:
                    flag=True
                    break
                else:
                    celula=self.labirinto[state[0]][state[1]-(i+1)]
                    if celula == "=" :
                        flag=True
            if flag==False:
                listactions.append("A")
            
                    
                    
                            
        if state[2]-1 > 0:
            newvelod=state[2]-1
            if state[3]=="N":
                flag=False
                for i in range(newvelod):
                    if state[0]-(i+1)<0:
                        flag=True
                        break
                    else:
                        celula=self.labirinto[state[0]-(i+1)][state[1]]
                        if celula == "=" :
                            flag=True
                if flag==False:
                    listactions.append("T")
                    
        
            if state[3]=="E":
                flag=False
                for i in range(newvelod):
                    if state[1]+(i+1)>=self.colunas:
                        flag=True
                        break
                    else:
                        celula=self.labirinto[state[0]][state[1]+(i+1)]
                        if celula == "=" :
                            flag=True
                if flag==False:
                    listactions.append("T")

            if state[3]=="S":
                
                flag=False
                for i in range(newvelod):
                    if state[0]+(i+1)>=self.linhas:
                        flag=True
                        break
                    else:
                        celula=self.labirinto[state[0]+(i+1)][state[1]]
                        if celula == "=" :
                            flag=True
                if flag==False:
                    listactions.append("T")
                    
            if state[3]=="O":
                flag=False
                for i in range(newvelod):
                    if state[1]-(i+1)<0:
                        flag=True
                        break
                    else:
                        celula=self.labirinto[state[0]][state[1]-(i+1)]
                        if celula == "=" :
                            flag=True
                if flag==False:
                    listactions.append("T")
            

        if state[2] == 1:
            
            listactions.append("T")
            
        new=sorted(listactions)
        
        
        return new
         

    def result (self, state, action):
        newstate=copy.deepcopy(state)
        
        if action=="E":
            if state[3]=="N":
                
                newstate[3]="O";
                
            elif state[3]=="E":
                
                newstate[3]="N";
                
            elif state[3]=="S":
                
                newstate[3]="E";

            elif state[3]=="O":
                
                newstate[3]="S";
            else:
                print("erro")
                
            
        elif action=="D":
            
            if state[3]=="N":
                newstate[3]="E";
                
            elif state[3]=="E":
                
                
                newstate[3]="S";
                
            elif state[3]=="S":
                newstate[3]="O";
                
            elif state[3]=="O":
                newstate[3]="N";
            else:
                print("erro")
        elif action=="A":
            
            if state[2]>self.vmax:
                newstate[2]=self.vmax
                if state[3]=="N":
                    self.labirinto[state[0]][state[1]]="."
                    
                    newstate[0]=newstate[0]-newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                
                elif state[3]=="E":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[1]=newstate[1]+newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                elif state[3]=="S":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[0]=newstate[0]+newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                elif state[3]=="O":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[1]=newstate[1]-newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                else:
                    print("erro")
            else:
                newstate[2]=newstate[2]+1
                if state[3]=="N":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[0]=newstate[0]-newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                elif state[3]=="E":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[1]=newstate[1]+newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                elif state[3]=="S":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[0]=newstate[0]+newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                elif state[3]=="O":
                    self.labirinto[state[0]][state[1]]="."
                    newstate[1]=newstate[1]-newstate[2]
                    self.labirinto[newstate[0]][newstate[1]]="o"
                else:
                    print("erro")
                
        elif action=="T":

            
            newstate[2]=newstate[2]-1
            
            if state[3]=="N":
                self.labirinto[state[0]][state[1]]="."
                newstate[0]=newstate[0]-newstate[2]
                self.labirinto[newstate[0]][newstate[1]]="o"
                
            elif state[3]=="E":
                self.labirinto[state[0]][state[1]]="."
                newstate[1]=newstate[1]+newstate[2]
                self.labirinto[newstate[0]][newstate[1]]="o"
                
            elif state[3]=="S":
                self.labirinto[state[0]][state[1]]="."
                
                newstate[0]=newstate[0]+newstate[2]
                self.labirinto[newstate[0]][newstate[1]]="o"
            elif state[3]=="O":
                self.labirinto[state[0]][state[1]]="."
                
                newstate[1]=newstate[1]-newstate[2]
                self.labirinto[newstate[0]][newstate[1]]="o"
            else:
                print("erro")
        else:
            print("erro")
            
        
        if action in self.actions(state):
            return newstate
        
            

    def goal_test (self, state):
        
        if state[0:2]==self.goal:
            if state[2]==0:
                return True
            else:
                return False
        else:
            return False
            

    def display (self, state):
        
        final=""
        labirintonew=copy.deepcopy(self.labirinto)
        for i in range(self.linhas):
            for j in range(self.colunas):
                if labirintonew[i][j]=="o":
                    if state[3]=="N":
                        labirintonew[i][j]="^"
                    elif state[3]=="E":
                        labirintonew[i][j]=">"
                    elif state[3]=="S":
                        labirintonew[i][j]="v"
                    elif state[3]=="O":
                        labirintonew[i][j]="<"
                    else:
                        print("erro")
                final=final+labirintonew[i][j]+" "
                
            final=final+"\n"
        
            
        return final

    def executa(self, state, actions_list, verbose=False):
        """Executa uma sequência de acções a partir do estado devolvendo o triplo formado pelo estado, 
        pelo custo acumulado e pelo booleano que indica se o objectivo foi ou não atingido. Se o objectivo 
        for atingido antes da sequência ser atingida, devolve-se o estado e o custo corrente.
        Há o modo verboso e o não verboso, por defeito."""
        cost = 0
        for a in actions_list:
            seg = self.result(state,a)
            cost = self.path_cost(cost,state,a,seg)
            state = seg
            obj = self.goal_test(state)
            if verbose:
                print('Ação:', a)
                print(self.display(state),end='')
                print('Custo Total:',cost)
                print('Atingido o objectivo?', obj)
                print()
            if obj:
                break
        return (state, cost, obj)

````

#### O construtor de `Labirinto`

O construtor da classe que irá desenvolver recebe como input informação em texto, referente ao labirinto inicial, com as suas dimensões, localização e orientação do automóvel e localização do objetivo, dados implicitamente. Além disso, também pode receber a velocidade máxima a que o automóvel se pode deslocar, por defeito está com o valor 3, i.e., o automóvel desloca-se 3 casas.

#### O método `actions`

As ações de mudança de orientação são duas: **Esquerda** e **Direita**, identificadas pelos símbolos "E" e "D", respetivamente. Estas ações mudam a orientação do automóvel em 90&#176; para a sua esquerda e para a sua direita, respetivamente. O agente só pode virar o volante se a sua velocidade for zero, permanecendo a zero depois da sua execução.

A ação **Acelera** incrementa a velocidade em 1 unidade e **Trava** decrementa-a em 1 unidade; em qualquer dos casos, o agente percorre um número de células iguais à sua nova velocidade. Estas duas ações são identificadas pelos símbolos "A" (acelera) e "T" (trava). Qualquer ação que leve o agente a bater contra uma parede é ilegal bem como reduzir a velocidade abaixo de zero. Também não é possível aumentar a velocidade acima de um máximo $v_{max}$, se a velocidade máxima tiver sido atingida e a ação for novamente Acelera, então a velocidade mantém-se constante e igual a $v_{max}$.

A lista de ações possíveis devolvidas por este método deverá estar **ordenada por ordem alfabética**.


Assim, para o exemplo acima, um plano possível mas não necessariamente o mais curto é:

direita + acelera + acelera + trava + trava + esquerda + acelera + acelera + trava + trava + esquerda + acelera + acelera + trava + trava

#### O método `result`

Repare que se um estado $s$ for objeto do método `result` ele deve permanecer exatamente igual, não sendo modificado pelo método. Deve ser gerado um estado completamente novo $s´$ e não alterar o estado $s$. Note que o teste de validade das ações deve ser feito no método `actions`, sendo ineficiente e redundante repeti-lo no método `result`.

#### Igualdade entre estados

É importante que dois estados exatamente com os mesmos valores nos seus atributos sejam considerados iguais mesmo que sejam objetos distintos!

Apenas precisa de ter em consideração situações iniciais de labirintos que sejam válidos. O principal objetivo desta avaliação é a formulação do problema num espaço de estados. Desenvolva o código assumindo sempre que o input do construtor da classe `Labirinto` é sempre válido.

#### O método `executa`

A função `executa` permite executar uma sequência de ações a partir de um determinado estado, devolvendo o estado que resulta da aplicação da sequência de ações, o custo total dessas ações e a indicação se o objetivo foi ou não satisfeito. Note que se o objetivo for atingido antes de aplicar todas as ações, o processo é interrompido sendo devolvido um triplo com a informação presente.

Há um modo verboso e não verboso, sendo este o modo por defeito. Se o colocar a True, após cada ação é apresentado o estado em modo txt, através do método `display`, juntamente com o custo total até ao momento e a indicação se o objetivo é atingido ou não. Serve para ajudar a testar o código.

Note que uma solução para o problema do labirinto representado acima (que não é ótima), com custo 15, seria:
```python
['D','A','A','T','T','E','A','A','T','T','E','A','A','T','T']
```
Permite testar, que partindo do estado inicial definido por defeito, e aplicando esta lista de ações atinge-se o estado final. Por exemplo, 
```python
p = Labirinto()
seq = ['D','A','A','T','T','E','A','A','T','T','E','A','A','T','T']
p.goal_test(p.executa(p.initial,seq)[0])
```
gera o seguinte output
```python
True
```
